```python
# Required imports
from collections import defaultdict, Counter
from statistics import mean

# Student identity
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

# Derived parameters
SEED_NUM = int(STUDENT_ID[-1])

ARTIST_POSITIONS = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
]

print("Setup complete!")
print("SEED_NUM:", SEED_NUM)
print("ARTIST_POSITIONS:", ARTIST_POSITIONS)
```

```python
student_names = [
    f"{LAST_NAME}_S{i}" for i in range(1, 6)
]

subjects = [
    "PROGRAMMING",
    "MATHEMATICS",
    "ELECTRONICS",
    "PHYSICS",
    "ENGLISH"
]

def score_category(score):
    if score is None:
        return "INCOMPLETE"
    elif score >= 90:
        return "EXCELLENT"
    elif score >= 80:
        return "VERY GOOD"
    elif score >= 70:
        return "GOOD"
    elif score >= 60:
        return "PASSING"
    else:
        return "NEEDS IMPROVEMENT"


performance_records = []

for i, student in enumerate(student_names):
    for j, subject in enumerate(subjects):
        artist_position = ARTIST_POSITIONS[
            (i + j) % len(ARTIST_POSITIONS)
        ]

        score = (
            55
            + (
                SEED_NUM * 7
                + artist_position * 3
                + i * 5
                + j * 4
            ) % 46
        )

        performance_records.append({
            "student": student,
            "subject": subject,
            "score": score,
            "category": score_category(score)
        })

# Required repeated student-subject record
performance_records.append(
    performance_records[0].copy()
)

# Required incomplete record
performance_records.append({
    "student": student_names[4],
    "subject": "ENGLISH",
    "score": None,
    "category": "INCOMPLETE"
})

print("=== GENERATED PERFORMANCE RECORDS ===")

for record in performance_records:
    print(record)
```

```python
student_subject_data = defaultdict(
    lambda: defaultdict(list)
)

for record in performance_records:
    student_subject_data[
        record["student"]
    ][
        record["subject"]
    ].append(record["score"])

print("=== ORGANIZED STUDENT DATA ===")

for student in sorted(student_subject_data):
    print("\n", student)

    for subject in sorted(student_subject_data[student]):
        print(
            subject,
            ":",
            student_subject_data[student][subject]
        )

performance_counts = Counter(
    (
        record["student"],
        record["subject"],
        record["score"]
    )
    for record in performance_records
)

repeated_performance = {
    record: count
    for record, count in performance_counts.items()
    if count > 1
}

incomplete_records = [
    record
    for record in performance_records
    if record["score"] is None
]

print("\n=== REPEATED RECORDS ===")
print(repeated_performance)

print("\n=== INCOMPLETE RECORDS ===")

for record in incomplete_records:
    print(record)
```

```python
print("=== STUDENT PERFORMANCE SUMMARIES ===")

for student in student_names:
    scores = [
        r["score"]
        for r in performance_records
        if r["student"] == student
        and r["score"] is not None
    ]

    print(
        f"{student}: "
        f"Valid scores={len(scores)}, "
        f"Average={mean(scores):.2f}, "
        f"Highest={max(scores)}, "
        f"Lowest={min(scores)}"
    )

print("\n=== SUBJECT PERFORMANCE SUMMARIES ===")

for subject in subjects:
    scores = [
        r["score"]
        for r in performance_records
        if r["subject"] == subject
        and r["score"] is not None
    ]

    print(
        f"{subject}: "
        f"Valid scores={len(scores)}, "
        f"Average={mean(scores):.2f}, "
        f"Highest={max(scores)}, "
        f"Lowest={min(scores)}"
    )
```

```python
category_counts = Counter(
    record["category"]
    for record in performance_records
)

ordered_performance = sorted(
    performance_records,
    key=lambda r: (
        r["student"],
        r["subject"]
    )
)

print("=== ORDERED PERFORMANCE RESULTS ===")

for record in ordered_performance:
    score_text = (
        "MISSING"
        if record["score"] is None
        else str(record["score"])
    )

    print(
        f'{record["student"]:<16} | '
        f'{record["subject"]:<14} | '
        f'{score_text:<7} | '
        f'{record["category"]}'
    )

print("\n=== FINAL PERFORMANCE SUMMARY ===")
print("Total records:", len(performance_records))
print(
    "Valid scores:",
    sum(r["score"] is not None for r in performance_records)
)
print("Incomplete records:", len(incomplete_records))
print("Repeated record combinations:", len(repeated_performance))
print("Performance categories:", dict(category_counts))
```
