```python
# Required imports
from collections import defaultdict, Counter
from statistics import mean

# Student identity
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

# Derived parameters
SEED_NUM = int(STUDENT_ID[-1])

ARTIST_POSITIONS = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
]

print("Setup complete!")
print("SEED_NUM:", SEED_NUM)
print("ARTIST_POSITIONS:", ARTIST_POSITIONS)
```

```python
from collections import defaultdict, Counter
from statistics import mean

test_sources = [f"{LAST_NAME}_T{i}" for i in range(1, 6)]

categories = [
    "TEMPERATURE",
    "VOLTAGE",
    "CURRENT",
    "PRESSURE",
    "SPEED"
]

test_records = []

for i, source in enumerate(test_sources):
    for j, category in enumerate(categories):
        artist_position = ARTIST_POSITIONS[
            (i + j) % len(ARTIST_POSITIONS)
        ]

        value = (
            SEED_NUM * 10
            + artist_position
            + i * 3
            + j * 2
        )

        test_records.append({
            "source": source,
            "category": category,
            "value": value
        })

# Additional measurements
test_records.extend([
    {
        "source": test_sources[0],
        "category": "TEMPERATURE",
        "value": SEED_NUM * 10 + ARTIST_POSITIONS[0] + 7
    },
    {
        "source": test_sources[1],
        "category": "VOLTAGE",
        "value": SEED_NUM * 10 + ARTIST_POSITIONS[1] + 11
    }
])

# Required repeated record
test_records.append(test_records[0].copy())

print("=== GENERATED TEST DATA ===")
for record in test_records:
    print(record)
```

```python
organized_tests = defaultdict(lambda: defaultdict(list))

for record in test_records:
    organized_tests[record["source"]][record["category"]].append(
        record["value"]
    )

print("=== ORGANIZED TEST DATA ===")

for source in sorted(organized_tests):
    print("\n", source)

    for category in sorted(organized_tests[source]):
        print(
            category,
            ":",
            organized_tests[source][category]
        )

record_counts = Counter(
    (r["source"], r["category"], r["value"])
    for r in test_records
)

repeated_records = {
    record: count
    for record, count in record_counts.items()
    if count > 1
}

distinct_records = set(
    (r["source"], r["category"], r["value"])
    for r in test_records
)

print("\n=== REPEATED RECORDS ===")
print(repeated_records)

print("\n=== DISTINCT DATA ===")
print("Distinct records:", len(distinct_records))
print("Total records:", len(test_records))
```

```python
values_by_category = defaultdict(list)
values_by_source = defaultdict(list)

for record in test_records:
    values_by_category[record["category"]].append(record["value"])
    values_by_source[record["source"]].append(record["value"])

print("=== SUMMARY BY CATEGORY ===")

for category in categories:
    values = values_by_category[category]

    print(
        f"{category}: "
        f"Count={len(values)}, "
        f"Minimum={min(values)}, "
        f"Maximum={max(values)}, "
        f"Mean={mean(values):.2f}, "
        f"Total={sum(values)}"
    )

print("\n=== SUMMARY BY SOURCE ===")

for source in test_sources:
    values = values_by_source[source]

    print(
        f"{source}: "
        f"Count={len(values)}, "
        f"Mean={mean(values):.2f}, "
        f"Total={sum(values)}"
    )
```

```python
ordered_records = sorted(
    test_records,
    key=lambda r: (
        r["source"],
        r["category"],
        r["value"]
    )
)

print("=== ORDERED TEST RESULTS ===")

for record in ordered_records:
    print(
        f'{record["source"]:<18} | '
        f'{record["category"]:<12} | '
        f'{record["value"]}'
    )

print("\n=== FINAL TEST DATA SUMMARY ===")
print("Number of sources:", len(test_sources))
print("Number of categories:", len(categories))
print("Total records:", len(test_records))
print("Distinct records:", len(distinct_records))
print("Repeated record combinations:", len(repeated_records))
print("Overall measurement total:", sum(r["value"] for r in test_records))
print(
    "Overall measurement mean:",
    f'{mean(r["value"] for r in test_records):.2f}'
)
```
