```python
# Required imports
from collections import defaultdict, Counter
from statistics import mean

# Student identity
LAST_NAME = "Sabariaga"
STUDENT_ID = "TUPM-26-1862"
FAVORITE_ARTIST = "MICO"

# Derived parameters
SEED_NUM = int(STUDENT_ID[-1])

ARTIST_POSITIONS = [
    ord(char.upper()) - ord("A") + 1
    for char in FAVORITE_ARTIST
    if char.isalpha()
]

print("Setup complete!")
print("SEED_NUM:", SEED_NUM)
print("ARTIST_POSITIONS:", ARTIST_POSITIONS)
```

```python
customers = [
    f"{LAST_NAME}_C{i}" for i in range(1, 6)
]

products = [
    "NOTEBOOK",
    "PEN",
    "USB",
    "HEADPHONES",
    "CALCULATOR"
]

product_prices = {
    "NOTEBOOK": 45.00,
    "PEN": 15.00,
    "USB": 250.00,
    "HEADPHONES": 350.00,
    "CALCULATOR": 180.00
}

transaction_records = []

for i, customer in enumerate(customers):
    for j, product in enumerate(products):
        artist_position = ARTIST_POSITIONS[
            (i + j) % len(ARTIST_POSITIONS)
        ]

        quantity = 1 + (
            (SEED_NUM + artist_position + i + j) % 4
        )

        price = product_prices[product]

        transaction_records.append({
            "customer": customer,
            "product": product,
            "quantity": quantity,
            "unit_price": price,
            "transaction_value": quantity * price
        })

# Required repeated customer-product transaction
transaction_records.append(
    transaction_records[0].copy()
)

print("=== GENERATED TRANSACTION RECORDS ===")

for record in transaction_records:
    print(record)
```

```python
transactions_by_customer = defaultdict(list)
transactions_by_product = defaultdict(list)

for record in transaction_records:
    transactions_by_customer[
        record["customer"]
    ].append(record)

    transactions_by_product[
        record["product"]
    ].append(record)

transaction_counts = Counter(
    (
        record["customer"],
        record["product"],
        record["quantity"],
        record["unit_price"]
    )
    for record in transaction_records
)

repeated_transactions = {
    transaction: count
    for transaction, count in transaction_counts.items()
    if count > 1
}

print("=== REPEATED TRANSACTIONS ===")
print(repeated_transactions)

print("\n=== TRANSACTION VALUES ===")

for i, record in enumerate(transaction_records, start=1):
    print(
        f'{i:02d}. {record["customer"]} | '
        f'{record["product"]} | '
        f'Quantity={record["quantity"]} | '
        f'Price={record["unit_price"]:.2f} | '
        f'Total={record["transaction_value"]:.2f}'
    )
```

```python
print("=== CUSTOMER SUMMARIES ===")

for customer in customers:
    records = transactions_by_customer[customer]

    total_spent = sum(
        record["transaction_value"]
        for record in records
    )

    total_items = sum(
        record["quantity"]
        for record in records
    )

    distinct_products = sorted({
        record["product"]
        for record in records
    })

    print(
        f"{customer}: "
        f"Transactions={len(records)}, "
        f"Items={total_items}, "
        f"Total spent={total_spent:.2f}, "
        f"Products={distinct_products}"
    )

print("\n=== PRODUCT SUMMARIES ===")

for product in products:
    records = transactions_by_product[product]

    total_quantity = sum(
        record["quantity"]
        for record in records
    )

    total_sales = sum(
        record["transaction_value"]
        for record in records
    )

    distinct_customers = sorted({
        record["customer"]
        for record in records
    })

    print(
        f"{product}: "
        f"Quantity sold={total_quantity}, "
        f"Transactions={len(records)}, "
        f"Total sales={total_sales:.2f}, "
        f"Customers={distinct_customers}"
    )
```

```python
ordered_transactions = sorted(
    transaction_records,
    key=lambda r: (
        r["customer"],
        r["product"],
        r["transaction_value"]
    )
)

print("=== ORDERED TRANSACTION RESULTS ===")

for record in ordered_transactions:
    print(
        f'{record["customer"]:<16} | '
        f'{record["product"]:<12} | '
        f'Quantity={record["quantity"]} | '
        f'Unit price={record["unit_price"]:.2f} | '
        f'Total={record["transaction_value"]:.2f}'
    )

print("\n=== FINAL TRANSACTION SUMMARY ===")

print("Total transaction records:", len(transaction_records))

print(
    "Distinct customers:",
    len({r["customer"] for r in transaction_records})
)

print(
    "Distinct products:",
    len({r["product"] for r in transaction_records})
)

print(
    "Repeated transaction combinations:",
    len(repeated_transactions)
)

print(
    "Total units sold:",
    sum(r["quantity"] for r in transaction_records)
)

print(
    "Overall transaction value:",
    f'{sum(r["transaction_value"] for r in transaction_records):.2f}'
)
```
